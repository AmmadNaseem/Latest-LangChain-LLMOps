```python
import os
from dotenv import load_dotenv, find_dotenv
_ = load_dotenv(find_dotenv())
if not os.getenv("GROQ_API_KEY"):
    raise ValueError("Add GROQ_API_KEY to the .env file before running this notebook.")

GROQ_MODEL = os.getenv("GROQ_MODEL", "qwen/qwen3.8-27b")
```

https://console.groq.com/docs/langchain

```python
# Groq chat model

from langchain_groq import ChatGroq

llmModel = ChatGroq(
    model=GROQ_MODEL,
    temperature=0.2,
    reasoning_effort="none",
)
```

```python
llmModel
```

```python
response  = llmModel.invoke("Tell me one fun fact about the Kennedy family.")
```

```python
print(response.content)
```

```python
for chunk in llmModel.stream(
    "Tell me one fun fact about the Kennedy family.i need detail response"
):
    print(chunk.content, end="", flush=True)
    # end=""      → Do not add a newline after each chunk
    # flush=True  → Display each chunk immediately

#Groq API → HTTPS response → SSE events → LangChain AIMessageChunk → terminal output
# Using end="" prevents that newline:
# flush=True
# Python may temporarily keep printed text in a memory buffer instead of immediately displaying it. 

# Client sends request
#         ↓
# Server keeps connection open
#         ↓
# Server sends event 1
# Server sends event 2
# Server sends event 3
#         ↓
# Server sends [DONE]
# Connection ends
```

#### Temperature: more or less creativity

```python
creativeLlmModel = ChatGroq(
    model=GROQ_MODEL,
    temperature=0.9,
    reasoning_effort="none",
)
```

| Value      | Behavior                                                        |
|------------|-----------------------------------------------------------------|
| `"none"`   | Direct answer; fastest and best for simple conversations.       |
| `"low"`    | Light reasoning with relatively low latency.                    |
| `"medium"` | Balanced reasoning and response speed.                          |
| `"high"`   | More reasoning for difficult tasks; slower and uses more tokens.|

```python
import os
from dotenv import load_dotenv
from langchain_groq import ChatGroq

load_dotenv(override=True)

GROQ_MODEL = os.getenv("GROQ_MODEL", "qwen/qwen3.8-27b")
GROQ_ALTERNATIVE_MODEL = os.getenv("GROQ_ALTERNATIVE_MODEL", "openai/gpt-oss-20b")
MAX_OUTPUT_TOKENS = int(os.getenv("GROQ_MAX_TOKENS", "256"))

primaryChatModel = ChatGroq(
    model=GROQ_MODEL,
    temperature=0.2,
    reasoning_effort="none",
    max_tokens=MAX_OUTPUT_TOKENS,
)

alternativeChatModel = ChatGroq(
    model=GROQ_ALTERNATIVE_MODEL,
    temperature=0.9,
    reasoning_effort="low",
    max_tokens=MAX_OUTPUT_TOKENS,
)

messages = [
    ("system", "You are an historian expert in the Kennedy family."),
    ("human", "Tell me one curious thing about JFK."),
]
primaryResponse = primaryChatModel.invoke(messages)
alternativeResponse = alternativeChatModel.invoke(messages)

print("Primary response:", primaryResponse.content)
print("Alternative response:", alternativeResponse.content)
```
